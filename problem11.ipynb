```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
titanic = sns.load_dataset('titanic')

print(titanic.head())
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='sex', y='age', hue='survived', data=titanic)

plt.title("Age Distribution by Gender and Survival (Boxplot)")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.violinplot(x='sex', y='age', hue='survived', data=titanic, split=True)

plt.title("Age Distribution by Gender and Survival (Violin Plot)")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```
