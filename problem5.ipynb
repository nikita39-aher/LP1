```python
import pandas as pd

# Load Titanic dataset
titanic_url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(titanic_url)

# Group by categorical variable (Sex)
grouped_stats = df.groupby('Sex')[['Age', 'Fare']].agg(['mean', 'median', 'min', 'max', 'std'])

print(grouped_stats)
```
