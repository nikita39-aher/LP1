```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
iris = sns.load_dataset('iris')

print(iris.head())
```

```python
print(iris.dtypes)
```

```python
iris.hist(figsize=(10,8))
plt.suptitle("Histograms of Iris Features")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=iris)

plt.title("Boxplot of Iris Features")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10,8))

axes[0,0].hist(iris['sepal_length'])
axes[0,0].set_title("Sepal Length")

axes[0,1].hist(iris['sepal_width'])
axes[0,1].set_title("Sepal Width")

axes[1,0].hist(iris['petal_length'])
axes[1,0].set_title("Petal Length")

axes[1,1].hist(iris['petal_width'])
axes[1,1].set_title("Petal Width")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10,8))

sns.boxplot(y=iris['sepal_length'], ax=axes[0,0])
axes[0,0].set_title("Sepal Length")

sns.boxplot(y=iris['sepal_width'], ax=axes[0,1])
axes[0,1].set_title("Sepal Width")

sns.boxplot(y=iris['petal_length'], ax=axes[1,0])
axes[1,0].set_title("Petal Length")

sns.boxplot(y=iris['petal_width'], ax=axes[1,1])
axes[1,1].set_title("Petal Width")

plt.tight_layout()
plt.show()
```
